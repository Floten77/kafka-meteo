Exercice 1 : Mise en place de Kafka et d’un producteur simple
1. Créez un topic Kafka nommé weather_stream .
2. Envoyez un message statique, par exemple :
{"msg": "Hello Kafka"}

Lancement de la commande suivante dans le container Kafka afin de créer le topic weather_stream:
docker exec kafka /opt/kafka/bin/kafka-topics.sh --create --topic weather_stream --bootstrap-server localhost:9092 --partitions 1 --replication-factor 1

```python
import json
from kafka import KafkaProducer

producer = KafkaProducer(
    bootstrap_servers="kafka:9092",
    value_serializer=lambda v: json.dumps(v).encode("utf-8")
)

message = {"msg": "Hello Kafka c'est un test "}

producer.send("weather_stream", value=message)
producer.flush()

print("Message envoyé :", message)

producer.close()
```
